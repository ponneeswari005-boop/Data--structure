```python
students = {}

n = int(input("Enter number of students: "))

for i in range(n):
    name = input("Enter Name: ")
    mark = int(input("Enter Mark: "))
    students[name] = mark

highest = max(students, key=students.get)

print("Student with Highest Mark:", highest)
print("Mark:", students[highest])
```
