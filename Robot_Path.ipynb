```python

A = (0, 0)
B = (5, 5)

obstacles = [(2, 2), (2, 3), (3, 3)]

x, y = A

while (x, y) != B:

    if (x + 1, y) not in obstacles and x < B[0]:
        x += 1
    elif (x, y + 1) not in obstacles and y < B[1]:
        y += 1
    else:
        print("Obstacle found!")
        break

    print((x, y))

if (x, y) == B:
    print("Reached Point B!")
```
